```python
import pandas as pd
import openpyxl

print("Pandas version:", pd.__version__)
print("Libraries imported successfully!")
```

```python
# Loading the dataset
file_path = "Data/Credit_Risk_Data.xlsx"

# Identifying available worksheets
excel_file = pd.ExcelFile(file_path)

# Displaying worksheet names
print(excel_file.sheet_names)
```

```python
# Loading both worksheets
# pd.read_excel() loads Excel worksheets into Pandas DataFrames
# 0 = first worksheet on the excel file; 1 = second worksheet
# 0 = Base Datasheet; 1 = Additional Datasheet
base_df = pd.read_excel(file_path, sheet_name=0)
additional_df = pd.read_excel(file_path, sheet_name=1)

# Checking dataset dimensions
# .shape tells us the number of rows and columns
print("Base Data:", base_df.shape)
print("Additional Data:", additional_df.shape)

# Previewing the first five rows using .head()
# display() shows the results as a formatte table in Jupyter
display(base_df.head())
```

```python
print("=== Dataset Overview & Data Quality Assessment ===\n")

# 1. Dataset dimensions
# shape = # of rows & columns
print("1. Dataset shape:", base_df.shape)

# 2. Column names
# .columns.tolist() = variable names
# \n = new line; \t = indentation; \n\n = two line breaks
print("\n2. Column Names:")
print(base_df.columns.tolist())

# 3. Data types and non-null counts
# .info() = data types & non-null counts
print("\n3. Dataset Information:")
base_df.info()

# 4. Missing values
# .isnull().sum() = missing values by variable
print("\n4. Missing Values:")
display(base_df.isnull().sum())

# 5. Duplicated rows
print("\n5. Duplicate Row:")
print(base_df.duplicated().sum())

# 6. Descriptive statistics
# .describe(include="all") = summary statistics for numerical & categorical variables)
print("\n6. Descriptive Statistics:")
display(base_df.describe(include="all"))
```

```python
print("=== Examining Unique Values in Categorical Columns ===")

# select_dtypes(include="object") = selects columns containing text, which are often categorical variables
# for col in ... = loops through those columns one at a time
# value_counts() = counts how often each value appears
# dropna=False = includes missing values in the counts
# f"\n{col}" = prints each column's name with a blank line (\n) before it
# .to_string = displays as plain text (just added to remove duplication of header)
for col in base_df.select_dtypes(include=["object","str"]).columns:
    print(f"\n{col}")
    print(base_df[col].value_counts(dropna=False).to_string(header=False))
```

```python
print("=== Credit Risk Distribution ===")

# value_counts() = counts the number of borrowers in each credit risk category
print(base_df["Credit Risk"].value_counts())

# Percentages
# normalize=True = calculates proportions instead of counts
# * 100 = converts proportions to percentages
print("\nPercentages (%):")
print((base_df["Credit Risk"].value_counts(normalize=True) * 100).round(2))
```

```python
print("=== Additional Dataset Overview ===")
# Determiing whether the additional dataset contains new borrowers, missing credit risk labels, or something else

# Check the number of rows and columns
print("\nDataset Shape:", additional_df.shape)

# Check the column names
print("\nColumn Names:")
print(additional_df.columns.tolist())

# Preview the first 5 rows
print("\nFirst 5 Rows:")
display(additional_df.head())

# Check missing values in the additional dataset
print("\nMissing Values:")
display(additional_df.isnull().sum())
```

```python
print("=== Comparing Dataset Columns ===")

# Check whether both datasets have the same column names
print("Same Columns:", base_df.columns.equals(additional_df.columns))

# Identify columns that appear in only one dataset
print("\nOnly in Base Dataset:")
print(base_df.columns.difference(additional_df.columns).tolist())

print("\nOnly in Additional Dataset:")
print(additional_df.columns.difference(base_df.columns).tolist())
```
