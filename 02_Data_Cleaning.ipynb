```python
import pandas as pd

file_path = "Data/Credit_Risk_Data.xlsx"

base_df = pd.read_excel(file_path, sheet_name=0)
additional_df = pd.read_excel(file_path, sheet_name=1)

print("=== Data Loaded Successfully ===")
print("Base Dataset:", base_df.shape)
print("Additional Dataset:", additional_df.shape)
```

```python
print("=== Data Type Comparison ===")

# .dtypes = shows the data type of each column
# pd.DataFrame() = creates a table from the information provided
# Compare the data types in both datasets

dtype_comparison = pd.DataFrame({
    "Base Dataset": base_df.dtypes,
    "Additional Dataset": additional_df.dtypes
})

display(dtype_comparison)
```

```python
print("=== Additional Dataset: Credit Risk Values ===")

# Check whether every Credit Risk value is missing
# .all() = returns True only if every value is missing
print(additional_df["Credit Risk"].isna().all())




print("\n=== Numerical Data Validation ===")

# Select numerical columns
numeric_cols = base_df.select_dtypes(include="number").columns

# Display minimum and maximum values for each numerical column
display(base_df[numeric_cols].agg(["min", "max"]))




print("\n=== Years Column Distribution ===")

# Count how often each value appears, sorted numerically
print(base_df["Years"].value_counts().sort_index())
```

```python
print("\n\n=== Additional Dataset: Numerical Data Validation ===")

# Select numerical columns, excluding the empty Credit Risk column
numeric_cols_add = additional_df.select_dtypes(include="number").columns.drop("Credit Risk")

# Display minimum and maximum values
display(additional_df[numeric_cols_add].agg(["min", "max"]))




print("\n=== Additional Dataset: Years Distribution ===")

# Count each value in ascending order
print(additional_df["Years"].value_counts().sort_index())

# Display borrowers with Years greater than 4
print("\nRecords with Years > 4:")
display(additional_df[additional_df["Years"] > 4])
```

```python
print("=== Categorical Value Comparison ===")

# Select text columns, excluding Credit Risk
categorical_cols = base_df.select_dtypes(include=["object", "str"]).columns.drop("Credit Risk")

# Compare unique values across both datasets
for col in categorical_cols:
    print(f"\n{col}")
    print("Base:", sorted(base_df[col].dropna().unique()))
    print("Additional:", sorted(additional_df[col].dropna().unique()))
```

```python
print("=== Standardizing Gender Values ===")

# .str.strip() = removes extra spaces
# .str.upper() = converts text to uppercase
base_df["Gender"] = base_df["Gender"].str.strip().str.upper()
additional_df["Gender"] = additional_df["Gender"].str.strip().str.upper()

# Check the results
print("\nBase Dataset:")
print(base_df["Gender"].value_counts(dropna=False))

print("\nAdditional Dataset:")
print(additional_df["Gender"].value_counts(dropna=False))
```

```python
print("=== Cross-Dataset Duplicate Check ===")

# Compare borrower information, excluding Credit Risk
feature_cols = base_df.columns.drop("Credit Risk")

# Find matching records between the two datasets
matching_rows = additional_df.merge(
    base_df[feature_cols],
    on=feature_cols.tolist(),
    how="inner"
)

print("Matching Records:", len(matching_rows))


print("\n=== Final Missing Value Check ===")

print("\nBase Dataset:")
print(base_df.isnull().sum())

print("\nAdditional Dataset:")
print(additional_df.isnull().sum())
```
